```python
def eyresi_klisis_2epipedwn(ABCd1,ABCd2):
    coosth=ABCd1[0]*ABCd2[0]+ABCd1[1]*ABCd2[1]+ABCd1[2]*ABCd2[2]
    coosth1=sqrt( (ABCd1[0]**2+ABCd1[1]**2+ABCd1[2]**2)*(ABCd2[0]**2+ABCd2[1]**2+ABCd2[2]**2) )
    coosth=abs(coosth/coosth1)

    if coosth==0:
        th=np.arccos(0)
        th=np.array((th))*180/np.pi #gia moires
    else:
        th=mth.acos(coosth)
        th=np.array((th))*180/np.pi # gia moires
    return th     
        

    
```
