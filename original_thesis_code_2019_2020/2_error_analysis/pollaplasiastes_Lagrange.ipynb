```python

n1=[n1x,n1y,n1z]

n2=[n2x,n2y,n2z]
p1=[p1x,p1y,p1z]
p2=[p2x,p2y,p2z]
p0=[p0x,p0y,p0z]
p=[px,py,pz] #zhtoymeno
w= lambdify([n1,n2,p1,p2,p0,l,m], p, 'numpy')
w=(px-p0x)**2+(py-p0y)**2+(pz-p0z)**2+l*px*n1x+l*py*n1y+l*pz*n1z-np.dot(np.dot(l,p1),n1)+ m*px*n2x+m*py*n2y+m*pz*n2z-np.dot(np.dot(m,p2),n2)
```

```python
## Continuing the recipe for Lagrange multipliers,
    #we compute the partial derivatives and set them to zero.
dpx=diff(w,px)
Edpx=Eq(dpx)

dpy=diff(w,py)
Edpy=Eq(dpy)

dpz=diff(w,pz)
Edpz=Eq(dpz)

dl=diff(w,l)
Edl=Eq(dl)
dm=diff(w,m)
Edm=Eq(dm)
```

```python

sol = solve((Edpx,Edpy,Edpz,Edl,Edm),(px,py,pz,l,m))

```
